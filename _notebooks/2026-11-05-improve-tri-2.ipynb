# Next Tri: Goals & Project Improvements

---

## 1. Project Improvements

For next trimester, I want to focus on making the PNEC website more useful, accessible, and user-friendly.

| Area | What I Want to Improve |
|---|---|
| Accessibility | Improve screen reader support and mobile compatibility. |
| Accuracy | Make sure emergency information comes from reliable sources. |
| User Experience | Make the website easier to navigate and understand. |
| Interactive Features | Add quizzes and activities about emergency preparedness. |
| Community Feedback | Get feedback from PNEC members and apply their suggestions. |
| Design | Improve the website's layout and overall organization. |

## 2. Personal Improvements

I also want to improve how I work individually and contribute to the team.

- **Time Management:** Stay organized and avoid leaving work until the last minute.
- **Communication:** Give teammates regular updates about my progress.
- **Initiative:** Take on more responsibility and contribute ideas independently.
- **Coding Skills:** Improve my backend development, debugging, and problem-solving.
- **Consistency:** Make steady progress on my assignments every week.
- **Teamwork:** Participate more in meetings and collaborate with teammates.

## 3. Goals for Next Tri

My main goals for the next trimester are:

1. Complete at least one major feature from planning to testing.
2. Make consistent GitHub commits and progress updates.
3. Test my code thoroughly before submitting pull requests.
4. Ask for feedback earlier in the development process.
5. Contribute more ideas during team meetings.
6. Focus on completing existing features before starting new ones.
7. Make sure my contributions benefit the PNEC community.

## 4. Measuring My Progress

By the end of next trimester, I want to be able to say that I have:

- Completed my assigned tasks on time.
- Made consistent contributions to GitHub.
- Successfully implemented and tested a major feature.
- Communicated effectively with my team.
- Used feedback to improve my work.
- Become more independent when solving coding problems.

## 5. Overall Reflection

This trimester helped me understand how much planning, communication, and teamwork go into developing a real project.

Next trimester, I want to become a more independent and reliable developer while helping our team make meaningful progress. I want to move beyond simply planning features and focus more on implementing, testing, and improving them.

Ultimately, I want our project to be more than just a functioning website. I want it to be something that is genuinely useful to the PNEC community.

### Main Focus for Next Tri

**Consistency, communication, initiative, stronger coding skills, and meaningful project improvements.**
