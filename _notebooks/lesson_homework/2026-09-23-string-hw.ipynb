## Popcorn Hack 1

```python
raw = "   Agent-Zero says: the drop is at MIDNIGHT!!!  "

clean = raw.strip()

start_idx = clean.find("MIDNIGHT")
end_idx = start_idx + len("MIDNIGHT")

raw_target = clean[start_idx:end_idx + 3]

target_clean = raw_target.replace("!", "")

code_word = target_clean.lower()
print("Extracted code word:", code_word)
```

## Popcorn Hack 3

```python
raw_transmission = "   2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87   "

# TODO 1: Use .strip() to remove outer whitespace, then .split(",") into fields.
segments = raw_transmission.strip().split(",")
print("Segments:", segments)

# TODO 2: Use indexing on segments to pull out date, agent, message code, and confidence.
date = segments[0]
agent = segments[1]
message_code = segments[2]
confidence_str = segments[3]
print(
    "Date:",
    date,
    "| Agent:",
    agent,
    "| Message code:",
    message_code,
    "| Confidence:",
    confidence_str,
)

# TODO 3: Use SLICING (not a method) on date to confirm it starts with the year "2026".
year_check = date[:4]
print("Year check:", year_check)

# TODO 4: Use .replace() to turn dashes in message_code into spaces.
message = message_code.replace("-", " ")

# TODO 5: Use .lower() and the `in` operator to check whether "dawn" appears in message.
found_dawn = "dawn" in message.lower()
print("Found 'dawn':", found_dawn)

# TODO 6: Use an f-string with "|" separators to build a Markdown table row.
markdown_row = f"| {date} | {agent} | {message} | {confidence_str}% |"
print("Markdown row:", markdown_row)

# TODO 7: Use an f-string to build a JSON-style line
json_line = (
    "{"
    + f'"date": "{date}", "agent": "{agent}", "message": "{message}", "confidence": {confidence_str}, "contains_dawn": {str(found_dawn).lower()}'
    + "}"
)
print("JSON line:", json_line)
```
