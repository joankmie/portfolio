```python
# Homework Hack - SFI Backend Validator
existing_spec_numbers = ["1.1", "2.1"]
valid_categories = ["Auto Racing", "Drag Racing"]

test_records = [
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.2",
        "effective_date": "Feb. 9, 2006",
    },
    {
        "product_name": "Replacement Flywheels",
        "category": "Street Car",
        "spec_number": "3.1",
        "effective_date": "Jan. 1, 2026",
    },
    {
        "product_name": "Existing Flywheel Record",
        "category": "Drag Racing",
        "spec_number": "1.1",
        "effective_date": "Jan. 1, 2026",
    },
]


def validate_sfi_record(record, existing_specs, valid_cats):
    """Validates an SFI candidate record based on complete criteria:

    - non-empty product name
    - accepted racing category
    - non-empty spec number
    - effective date present
    - non-duplicate spec number
    """
    has_product = bool(record.get("product_name", "").strip())
    is_valid_cat = record.get("category") in valid_cats
    has_spec = bool(record.get("spec_number", "").strip())
    has_date = bool(record.get("effective_date", "").strip())
    is_duplicate = record.get("spec_number") in existing_specs

    is_accepted = (
        has_product and is_valid_cat and has_spec and has_date and not is_duplicate
    )

    reasons = []
    if not has_product:
        reasons.append("Missing product name")
    if not is_valid_cat:
        reasons.append(f"Invalid category '{record.get('category')}'")
    if not has_spec:
        reasons.append("Missing spec number")
    if not has_date:
        reasons.append("Missing effective date")
    if is_duplicate:
        reasons.append(
            f"Duplicate spec number '{record.get('spec_number')}' already exists"
        )

    return is_accepted, reasons


print("=== SFI BACKEND VALIDATION RESULTS ===")
for i, candidate in enumerate(test_records, start=1):
    accepted, issues = validate_sfi_record(
        candidate, existing_spec_numbers, valid_categories
    )
    status = "ACCEPTED" if accepted else "REJECTED"

    print(f"\nCandidate Record {i}: {candidate.get('product_name')}")
    print(f"Decision: {status}")
    if not accepted:
        print("Rejection Reasons:", ", ".join(issues))
```
