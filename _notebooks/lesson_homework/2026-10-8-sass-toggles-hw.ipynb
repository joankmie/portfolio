```python

<div id="preparedness-dashboard">
  <h2>Emergency Preparedness Dashboard</h2>
  <p>Turn features on or off using the switches.</p>

  <!-- Toggle 1 controls whether safety tips are visible. -->
  <label class="ocs__toggle ocs__toggle--green">
    <input id="safety-toggle" class="ocs__toggle-input" type="checkbox" checked>
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Safety Tips</span>
  </label>

  <!-- Toggle 2 controls the emergency checklist. -->
  <label class="ocs__toggle ocs__toggle--green">
    <input id="checklist-toggle" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Emergency Checklist</span>
  </label>

  <!-- Toggle 3 controls the sample weather alert. -->
  <label class="ocs__toggle ocs__toggle--green">
    <input id="weather-toggle" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Weather Alerts</span>
  </label>

  <hr>

  <h3>Enabled Features: <output id="toggle-count" aria-live="polite">1</output> / 3</h3>

  <section id="safety-content">
    <h3>Safety Tips</h3>
    <ul>
      <li>Keep an emergency kit with water and essential supplies.</li>
      <li>Know at least two ways to leave your home safely.</li>
      <li>Keep important emergency contact information accessible.</li>
    </ul>
  </section>

  <section id="checklist-content" hidden>
    <h3>Emergency Checklist</h3>
    <ul>
      <li>Prepare drinking water and nonperishable food.</li>
      <li>Pack a flashlight and extra batteries.</li>
      <li>Write down important emergency contacts.</li>
      <li>Choose a family meeting location.</li>
    </ul>
  </section>

  <section id="weather-content" hidden>
    <h3>Sample Weather Alert</h3>
    <p>Check official local sources for current weather alerts and instructions.</p>
  </section>
</div>

<script>
  // Store the toggles and the sections they control.
  const safetyToggle = document.getElementById("safety-toggle");
  const checklistToggle = document.getElementById("checklist-toggle");
  const weatherToggle = document.getElementById("weather-toggle");

  const safetyContent = document.getElementById("safety-content");
  const checklistContent = document.getElementById("checklist-content");
  const weatherContent = document.getElementById("weather-content");
  const toggleCount = document.getElementById("toggle-count");

  // The checked property is true when a checkbox is turned on.
  // The hidden property determines whether a section is displayed.
  function updateDashboard() {
    safetyContent.hidden = !safetyToggle.checked;
    checklistContent.hidden = !checklistToggle.checked;
    weatherContent.hidden = !weatherToggle.checked;

    // Count the enabled toggles each time a setting changes.
    const enabledCount = [
      safetyToggle,
      checklistToggle,
      weatherToggle
    ].filter(toggle => toggle.checked).length;

    toggleCount.textContent = enabledCount;
  }

  // Listen for changes so each switch updates its feature immediately.
  safetyToggle.addEventListener("change", updateDashboard);
  checklistToggle.addEventListener("change", updateDashboard);
  weatherToggle.addEventListener("change", updateDashboard);

  // Initialize the dashboard using the starting checkbox states.
  updateDashboard();
</script>
```
