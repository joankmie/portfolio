## Design Thinking

- **Player Need:** Accessibility options that reduce physical discomfort (motion sickness or vestibular distress) caused by fast visual animations.

- **My Goal:** Provide a clear, customizable setting in the game system that allows players to toggle reduced-motion effects without altering core gameplay mechanics or star counts.

- **Setting/Representation Considered:** Represented reduced_motion as a Boolean (True/False) value so that render pipelines can easily check it before running screen-shake or particle animations.

- **Prototype Built:** Initialized reduced_motion = True alongside string and integer values representing game parameters and star tracking.

- **Test / Revision:** Verified that changing state variables (like bonus or reduced_motion) operates independently of other state values, ensuring configuration settings do not interfere with score tracking logic.

- **Why Reduced Motion Should Be a Choice:** Players have varying physiological sensitivities. Forcing motion-heavy effects can cause dizziness or headaches, making the game unplayable for some. Allowing it to be a choice ensures inclusivity and equal access for all players.
