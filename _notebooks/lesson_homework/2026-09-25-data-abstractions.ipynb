### Complexity & Abstraction Explanation
Without list abstraction, tracking 100 levels would require 100 separate variables (`level1_stars`, `level2_stars`, etc.) and a manual sum statement containing 100 additions (`total = level1_stars + level2_stars + ...`).

Using a single list combined with a `for` loop reduces all 100 variables down to one list data structure and allows the same 2-line loop traversal to calculate the total automatically—regardless of whether the game has 3 levels or 100 levels.

## Design Thinking
- **Player/Maker Need:** Game creators need a manageable way to track star counts for an expanding list of levels without rewriting tracking logic as levels are added or modified. Players need clear information about their progress.

- **Goal:** Create a dynamic system to store, update, and sum level star counts automatically.

- **Representation Considered:** Individual scalar variables vs. a single list. We selected a list because index-based lookup and loop traversals scale effortlessly.

- **Prototype:** `level_stars = [12, 18, 10]` with a `for` loop traversal for summation.

- **Test & Revision:** During edge-case testing with an empty list (`[]`), attempting `level_stars[1] = level_stars[1] + 5` caused an IndexError. Revision: Wrapped element-specific updates in a conditional length check (`if len(level_stars) >= 2:`) before attempting index-based access.

- **UX Suggestion:** Display level statistics using 1-based labels (e.g.,  `"Level 1"`, `"Level 2"`) in the user interface, hiding zero-based indexing (`0`, `1`) so players do not get confused by programming representations.
