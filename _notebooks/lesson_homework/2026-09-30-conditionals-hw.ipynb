## Popcorn Hack 2

1. The trainer tells the person: "Be a little careful, but this is probably fine."

2. The branches that can never run are `elif risk_score >= 6:` and `elif risk_score >= 9:` or anything greater than 3.

```python
risk_score = 9

if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

## Homework Hack

```python
message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False

risk_score = 0

if "gift card" in message:
    risk_score += 4

if amount_requested > 100:
    risk_score += 3

if not known_contact:
    risk_score += 2

if risk_score >= 7:
    print(f"Risk Score: {risk_score} - CRITICAL: High-risk scam detected! Do not reply.")
elif risk_score >= 4:
    print(f"Risk Score: {risk_score} - WARNING: Suspicious activity detected. Exercise caution.")
elif risk_score >= 2:
    print(f"Risk Score: {risk_score} - NOTICE: Unknown sender, but request seems low risk.")
else:
    print(f"Risk Score: {risk_score} - SAFE: Message appears to be completely safe.")
```
