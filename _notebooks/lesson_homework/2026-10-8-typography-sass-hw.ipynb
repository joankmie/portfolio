## Popcorn Hack

**Explanation of Refactoring:**

Removed: All `<style>` blocks and non-semantic CSS classes (`.large-title,` .`sub-title`, `.normal`).

**Semantic Tags Applied:**

Replaced `<div class="large-title">` with `<h2>` (Section Title).

Replaced `<span class="sub-title">` with `<h3>` (Supporting Heading).

Replaced `<div class="normal">` with `<p>` (Body Paragraph).

OCS Grammar Applied: Used `ocs_section-title`, `ocs_lead`, and `ocs_text`.

**Code:**

```python
%%html
<!-- UI_RUNNER: Typography Popcorn Hack -->
<h2 class="ocs_section-title">Why Cats Purr</h2>
<h3 class="ocs_lead">A Sign of Happiness</h3>
<p class="ocs_text">Cats usually purr when they feel safe, cozy, and ready for a nap.</p>
```
