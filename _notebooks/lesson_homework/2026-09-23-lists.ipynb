### Popcorn Hack 3

````python

seasons = ["winter", "summer", "fall", "summer", "spring", "summer", "fall"]   # use real class data

frequency = {}
for season in seasons:
    if season in frequency:
        frequency[season] += 1
    else:
        frequency[season] = 1

popular = []
for freq in frequency:
    if frequency[freq] >= 2:
        popular.append(freq)

print("Popular Seasons (appearing 2 or more times):", popular)
````

### Homework

````python

votes = ["chips", "cookies", "chips", "fruit", "cookies", "chips",
         "pretzels", "cookies", "chips", "fruit", "chips", "cookies"]

prices = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

number_of_votes = len(votes)
print("Number of votes cast:", number_of_votes)
print("Last vote cast:", votes[-1])

total = 0
for price in prices:
    total += price
average = total / len(prices)
print("Total: ", int(total))
print("Average: ", int(average))

frequency = {}

for vote in votes:
    if vote in frequency:
        frequency[vote] += 1
    else:
        frequency[vote] = 1
print("Frequency of votes:", frequency)

snacks_with_3_or_more_votes = []
for snack, count in frequency.items():
    if count >= 3:
        snacks_with_3_or_more_votes.append(snack)
print("Snacks with 3 or more votes:", snacks_with_3_or_more_votes)
````
