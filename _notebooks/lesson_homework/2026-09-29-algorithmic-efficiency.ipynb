**Prediction for n = 16:**

For `n = 16` books, worst-case linear search checks will double from 8 to 16 (`O(n)`), while binary search checks will only increase by 1, going from 4 to 5 (`log 2 (16) + 1 = 5`).

Actual Output will be shown after the code block.

```python
import math

n = 16
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1

print(f"Catalog size (n): {n}")
print(f"Linear Search Worst-Case Checks: {linear_checks}")
print(f"Binary Search Worst-Case Checks: {binary_checks}")
```

Actual Output is below code cell


```python
n = 12
visits = n
pairs = n * n

print(f"Book Count (n): {n}")
print(f"Single Visits (Linear): {visits}")
print(f"Ordered Pairs (Quadratic): {pairs}")
```

### Library Search Recommendation (n = 1,000)

1. **For 1,000 Sorted Books:**

   - Linear Search Worst-Case: 1,000 checks.

   - Binary Search Worst-Case: `floor(log 2(1000)) + 1 = 10` checks.

   - Recommendation: Binary search is overwhelmingly recommended, taking only 10 checks compared to 1,000.

2. **Impact of Unsorted Catalog or Memory Constraints:**

   - If Unsorted: Binary search requires data to be sorted. Sorting the catalog upfront takes time (`O(n / log n)` work). If we only do a single search, linear search (1,000 checks) is faster than sorting + binary search. However, if doing repeated searches, sorting once pay off quickly.

   - If Extra Memory is Required: Creating and storing an extra sorted copy of the catalog uses additional space/memory (`O(n)` space complexity). If system storage is severely limited, there is a space/time tradeoff to evaluate.
